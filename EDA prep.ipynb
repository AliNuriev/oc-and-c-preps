```python
df_laptops = pd.read_csv("laptop_price.csv", encoding = "latin-1")

conditions = [
    df_laptops["Price_euros"] > 3000,
    (df_laptops["Price_euros"] <= 3000) & (df_laptops["Price_euros"] > 2000), 
    (df_laptops["Price_euros"] <= 2000) & (df_laptops["Price_euros"] > 800),
    (df_laptops["Price_euros"] <= 800)
]

values = ["Too Expensive", "Expensive", "Affordable", "Cheap"]
```

```python
df_laptops["price_tiers"] = np.select(conditions, values, default = "Unknown")
```

```python
df_laptops.head()
```

```python
df_laptops.value_counts("price_tiers")
```

```python
condition_screen = [
    df_laptops["Inches"] > 16,
    (df_laptops["Inches"] <= 16) & (df_laptops["Inches"] >14),
    (df_laptops["Inches"] <= 14) & (df_laptops["Inches"] >12),
    df_laptops["Inches"] <= 12
]

values_screen = ["Too Big", "Big", "Small", "Too Small"] 
np.select(condition_screen, values_screen, default = "Unknown")
```

```python
df_laptops["Screen size tiers"] = np.select(condition_screen, values_screen, default = "Unknown")
```

```python
df_laptops.head()
```

```python
df_laptops.value_counts("Screen size tiers")
```
