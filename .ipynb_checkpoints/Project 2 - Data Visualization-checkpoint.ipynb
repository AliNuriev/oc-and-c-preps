```python
import pandas as pd
```

```python
df_population_raw = pd.read_csv("population_total.csv")
```

```python
df_population_raw.dropna(inplace = True)
```

```python
df_population_raw
```

```python
df_pivot = df_population_raw.pivot(columns = "country", index = "year", values = "population")
```

```python
df_pivot
```

```python
df_pivot = df_pivot[["United States", "Brazil", "China", "Russia", "Ukraine"]]
df_pivot
```

```python
df_pivot.plot(kind = "line", xlabel = "Year", ylabel = "Population", title = "Population growth", figsize = (10, 6))
```

```python
df_2020 = df_pivot[df_pivot.index.isin([2020.0])]
```

```python
df_2020
```

```python
df_2020 = df_2020.T
```

```python
df_2020.plot(kind = "bar", color = "black", xlabel = "Country", ylabel = "Population", title = "Population in 2020")
```

```python
df_pivot_1990_2020 = df_pivot[df_pivot.index.isin([1990, 2000, 2010, 2020])]
```

```python
df_pivot_1990_2020.plot(kind = "bar", xlabel = "Country", ylabel = "Population", title = "Grouped by year population change in different countries", figsize = (10, 6))
```

```python
df_2020.rename(columns = {2020: "2020"}, inplace = True)
```

```python
df_2020.plot(kind = "pie", y = "2020")
```
