```python
import pandas as pd
import cufflinks as cf
from IPython.display import display, HTML
import plotly.express as px

cf.set_config_file(sharing = "public", theme = "ggplot", offline = True)
```

```python
cf.getTheme()
```

```python
df_population_raw = pd.read_csv("population_total.csv")
df_population_raw.dropna(inplace = True)
df_pivot = df_population_raw.pivot(index = "year", columns = "country", values = "population")
```

```python
df_pivot = df_pivot[["United States", "India", "China", "Brazil", "Indonesia"]]
```

```python
df_pivot
```

# 1 Lineplot

```python
fig = px.line(
    df_pivot,
    labels={"year": "Year", "value": "Population", "country": "Country"},
    title="Population by country",
    width = 1100,
    height = 800
)
fig.show()
```

# 2 Barplot

```python
df_pivot_2020 = df_pivot[df_pivot.index.isin([2020])]
df_pivot_2020 = df_pivot_2020.T
```

```python
df_pivot_2020
bar = px.bar(df_pivot_2020,
            labels = {"country": "Country", "values":"Population, bln", "year": "Year"},
            title = "Population of 5 countries in 2020",
            width = 1100,
            height = 800)
```

```python
bar
```
